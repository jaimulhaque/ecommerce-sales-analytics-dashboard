# 📊 E-Commerce Sales Analytics Dashboard

## Day 1: Data Exploration

### Project Overview

This project focuses on analyzing e-commerce sales data to discover business insights using SQL, Python, and Power BI.

The workflow includes:

- Data Understanding
- Data Cleaning
- Exploratory Data Analysis (EDA)
- SQL Business Analysis
- Power BI Dashboard Development

---

## Objective

The objective of this analysis is to understand:

- Sales performance
- Customer behavior
- Product performance
- Regional trends
- Profitability patterns

---

## Tools Used

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Jupyter Notebook
